```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_sales_dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.drop_duplicates()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='X', data=df)
plt.title("Distribution of X")
plt.show()
```

```python
gender = df['X'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender,
        labels=gender.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Distribution of X")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Y'], bins=10, kde=True)
plt.title("Y Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['RSS - A'], fill=True)
plt.title("Distribution of RSS - A")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['RSS - A'])
plt.title("Box Plot of RSS - A")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='X', y='RSS - A', data=df, alpha=0.6)
plt.title("Scatter Plot of RSS - A vs X")
plt.xlabel("X")
plt.ylabel("RSS - A")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y='Total Amount', data=df)
plt.title("Box Plot of Total Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Select only numeric columns for correlation
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of Numeric Variables")
plt.show()
```

```python
sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='X', y='RSS - A', data=df, alpha=0.5)
plt.title("Strip Plot of RSS - A by X")
plt.show()
```

```python
print(sns.color_palette("deep"))
print(sns.color_palette("Set1"))
print(sns.color_palette("Set2"))
print(sns.color_palette("viridis"))
print(sns.color_palette("plasma"))
print(sns.color_palette("magma"))
print(sns.color_palette("Blues"))
print(sns.color_palette("Greens"))
print(sns.color_palette("Reds"))
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='X', y='RSS - A', data=df, alpha=0.8)
plt.title("Scatter Plot of RSS - A vs X")
plt.xlabel("X")
plt.ylabel("RSS - A")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
